Q.1 Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.

Driver- The main process that creates the SparkSession, converts code into tasks, and coordinates execution.
Cluster Manager - Allocates resources across the cluster.
Executor - Worker processes that execute tasks and store data in memory or disk.

Q.2 How does Spark’s Lazy Evaluation strategy improve performance when chain-processing large datasets?

Spark does not execute transformations immediately. Instead, it builds a DAG (Directed Acyclic Graph) of operations and executes them only when an action is called. This allows Spark to optimize the execution plan, reduce unnecessary data movement, and combine operations for better performance.

Q.7 How does Spark use the Lineage Graph (DAG) for fault tolerance.

Spark records all transformations in a lineage graph (DAG). If a worker node fails and data is lost, Spark recomputes only the missing partitions using the lineage information instead of reprocessing the entire dataset.

Q.9 Explain Predicate Pushdown in Parquet.

Predicate Pushdown allows Spark to send filter conditions directly to the Parquet reader. Only rows matching the filter are read from disk, reducing disk I/O, network transfer, and memory consumption.

Q.13 Difference between Client Mode and Cluster Mode.

Client Mode - \
1. Driver runs on the client machine
2. Client must remain connected.

Cluster Mode - \
1. Driver runs inside the cluster.
2. Client can disconnect after job submission.

Q.15 Why use .show(5) instead of .collect() on a multi-terabyte dataset.


1. show(5) displays only a small sample of rows.
2. collect() brings all data from executors to the driver's memory.
3. On a multi-terabyte dataset, .collect() can cause out-of-memory errors, network congestion, and application failure.
Therefore, .show(5) is much safer for data exploration.
