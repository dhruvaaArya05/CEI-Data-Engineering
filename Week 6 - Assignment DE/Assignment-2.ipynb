Read CSV with Schema Inference

```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Assignment") \
    .getOrCreate()

df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("source.csv")
```

Explore Data

```python
df.show(5);
```

Filter Electronics Products

```python
electronics_df = df.select(
    "product_id",
    "price"
).filter(df.category == "Electronics")

electronics_df.show()
```

Rename Column and Cast Data Type

```python
from pyspark.sql.functions import col

df_revised = df \
    .withColumnRenamed("old_name", "new_name") \
    .withColumn(
        "price",
        col("price").cast("double")
    )
```

Add New Column

```python
from pyspark.sql.functions import col

df_revised = df_revised.withColumn(
    "final_price",
    col("base_price") * 1.18
)
```
